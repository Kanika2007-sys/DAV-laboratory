```python
import pandas as pd
import numpy as np
import scipy.stats as stats

# Define column names explicitly
col_names = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome']

# Load dataset with headers specified
uci_df = pd.read_csv('https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv', names=col_names)
pima_df = uci_df.copy()

# Define numerical columns for analysis
numeric_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']

# Calculate Univariate Statistics for UCI Diabetes Dataset
uci_stats = pd.DataFrame({
    'Mean': uci_df[numeric_cols].mean(),
    'Median': uci_df[numeric_cols].median(),
    'Mode': uci_df[numeric_cols].mode().iloc[0],
    'Variance': uci_df[numeric_cols].var(),
    'Standard Deviation': uci_df[numeric_cols].std(),
    'Skewness': uci_df[numeric_cols].skew(),
    'Kurtosis': uci_df[numeric_cols].kurtosis()
})

# Calculate Univariate Statistics for Pima Indians Diabetes Dataset
pima_stats = pd.DataFrame({
    'Mean': pima_df[numeric_cols].mean(),
    'Median': pima_df[numeric_cols].median(),
    'Mode': pima_df[numeric_cols].mode().iloc[0],
    'Variance': pima_df[numeric_cols].var(),
    'Standard Deviation': pima_df[numeric_cols].std(),
    'Skewness': pima_df[numeric_cols].skew(),
    'Kurtosis': pima_df[numeric_cols].kurtosis()
})

print("--- UCI Diabetes Dataset Univariate Analysis ---")
print(uci_stats)

print("\n--- Pima Indians Diabetes Dataset Univariate Analysis ---")
print(pima_stats)
```
