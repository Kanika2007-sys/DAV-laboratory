```python
import pandas as pd

# 1. Read CSV Data from Web (Iris Dataset - UCI Machine Learning Repository)
text_df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv')

# 2. Read Excel Data from Web (Financial Sample Dataset)
# Replace line 7 with this working link
excel_url = 'https://raw.githubusercontent.com/Premalatha-success/Datasets/main/Global%20Superstore.xlsx'
excel_df = pd.read_excel(excel_url, sheet_name='Orders')

# 3. Read Web Data (Country Demographics / World Data)
web_df = pd.read_csv('https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv')

# --- Display original heads ---
print("--- 1. CSV Data (Iris) ---")
print(text_df.head(), "\n")

print("--- 2. Excel Data (Superstore Orders) ---")
print(excel_df[['Order ID', 'Category', 'Sub-Category', 'Sales', 'Profit']].head(), "\n")

print("--- 3. Web Data (Countries) ---")
print(web_df.head(), "\n")

# --- Handle missing values ---
text_df.ffill(inplace=True)
excel_df.bfill(inplace=True)
web_df.dropna(inplace=True)

# --- Save processed data to output files ---
text_df.to_csv('processed_text.csv', index=False)
excel_df.to_excel('processed_excel.xlsx', index=False)

print("SUCCESS: Data read, cleaned, and exported successfully!")
```
